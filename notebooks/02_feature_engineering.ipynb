```python
#import các thư viện cần thiết
import pandas as pd
import numpy as np
import re
```

```python
#đọc file csv
df = pd.read_csv("../data/Amazon-Products_processed.csv")
df_feature = df.copy()
```

```python
df.info()
```

```python
#khởi tạo list
text_columns = [
    "name",
    "main_category",
    "sub_category"
]
```

```python
#đưa vể lowercase
for col in text_columns:
    df_feature[col] = (
        df_feature[col]
        .astype("string")
        .str.lower()
    )
```

```python
#xóa khoảng trắng thừa
for col in text_columns:
    df_feature[col] = (
        df_feature[col]
        .str.strip()
        .str.replace(r"\s+", " ", regex=True)
    )
```

```python
#xóa kí tự đặc biệt
def clean_text(text):
    text = str(text).lower()

    # Xóa các ký hiệu thương mại
    text = re.sub(r"[®©™]", "", text)

    # Chuẩn hóa khoảng trắng
    text = re.sub(r"\s+", " ", text)

    return text.strip()

for col in text_columns:
    df_feature[col] = df_feature[col].apply(clean_text)
```

```python
#tạo product_id
df_feature = df_feature.reset_index(drop=True)
df_feature["product_id"] = df_feature.index
df_feature[
    [
        "product_id",
        "name"
    ]
].head()
```

```python
#tạo product_text
df_feature["product_text"] = (
    "Product: " + df_feature["name"] +
    ". Main category: " + df_feature["main_category"] +
    ". Sub category: " + df_feature["sub_category"]
)

df_feature[
    [
        "product_id",
        "product_text"
    ]
].head()
```

```python
df_feature.to_csv(
    "../data/Amazon-Products_feature.csv",
    index=False
)
```
