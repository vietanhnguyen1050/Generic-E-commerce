```python
#import các thư viện cần thiết
import numpy as np
import faiss
```

```python
# Đọc embedding
embeddings = np.load("../models/embeddings.npy").astype("float32")

# Chuẩn hóa
faiss.normalize_L2(embeddings)

# Tạo FAISS Index
dimension = embeddings.shape[1]
index = faiss.IndexFlatIP(dimension)

# Thêm vector vào index
index.add(embeddings)

# Lưu index
faiss.write_index(index, "../models/product_index.faiss")

print(f"Đã lưu {index.ntotal} vector vào FAISS.")
```
